```python
import os
import sys
import importlib
module_path = os.path.abspath(os.path.join('..'))
if module_path not in sys.path:
    sys.path.append(module_path)
ocsv = importlib.import_module("Simulation and BeamLib.optris_csv")
import matplotlib.pyplot as plt
from scipy.signal import medfilt

TIME_CUTOFF = 2000
filepath = "C:/Users/ssuub/Desktop/EDET80k_Damage/Lasing Analysis/apps/Simulation and BeamLib/Caleb/New PC Data/10V_Data/2026/Al10/HeatingCalibration.dat"
kernel_size = 10001

keys = ["TL", "TR", "ML", "MR", "BL", "BR"]
styles = ["-", "--", "-", "--", "-", "--"]
colors = ["red", "red", "green", "green", "blue", "blue"]

data = ocsv.OptrisDataset(filepath, ";").build_array_data()
data = data[data["time"] <= TIME_CUTOFF]
for i, key in enumerate(keys):
    temp = medfilt(data[key], kernel_size = kernel_size)
    plt.plot(data["time"], temp, label = key, linestyle = styles[i], color = colors[i], linewidth = 2)

plt.xlabel("Time (s)")
plt.ylabel("Temperature (°C)")
plt.title("Temperature of Various Points on Sample During Heating")
plt.legend()
plt.show()
```

```python
from scipy.optimize import curve_fit
import numpy as np

# Define thermal conduction model (1D heat diffusion)
def thermal_model(t, t0, Ts, tau):
    """
    t0: time offset
    Ts: steady state temperature
    tau: time constant for heat diffusion
    """
    T0 = data[key][0]  # Initial temperature from data
    return T0 + (Ts - T0) * (1 - np.exp(-(t - t0) / tau))


# Fit model to each dataset
fitted_params = {}
for key in keys:
    filtered_temp = medfilt(data[key], kernel_size=kernel_size)
    
    # Initial parameter guesses
    p0 = [filtered_temp[0], 10, 500]
    
    try:
        popt, pcov = curve_fit(thermal_model, data["time"], filtered_temp, p0=p0, maxfev=5000)
        fitted_params[key] = {'params': popt, 'covariance': pcov}
        print(f"{key}: T0={popt[0]:.2f}, k={popt[1]:.4f}, tau={popt[2]:.2f}")
        
    except Exception as e:
        print(f"Failed to fit {key}: {e}")

# Plot fits against data
#plt.figure(figsize=(12, 6))
for i, key in enumerate(keys):
    filtered_temp = medfilt(data[key], kernel_size=kernel_size)
    plt.plot(data["time"], filtered_temp, label=f"{key} (data)", 
             linestyle=styles[i], color=colors[i], linewidth=2, alpha=0.7)
    
    if key in fitted_params:
        fitted_curve = thermal_model(data["time"], *fitted_params[key]['params'])
        plt.plot(data["time"], fitted_curve, linestyle=':', color=colors[i], linewidth=2)

plt.xlabel("Time (s)")
plt.ylabel("Temperature (°C)")
plt.title("Thermal Conduction Model Fits")
plt.legend()
plt.show()
```

```python
KELVIN_OFFSET = 273.15

def intensity(temperature):
    return (temperature + KELVIN_OFFSET) ** 4
def temperature(intensity):
    return intensity ** 0.25 - KELVIN_OFFSET

tdata = (intensity(medfilt(data["TR"], kernel_size = kernel_size)) / intensity(medfilt(data["TL"], kernel_size = kernel_size))) ** 0.25
mdata = (intensity(medfilt(data["MR"], kernel_size = kernel_size)) / intensity(medfilt(data["ML"], kernel_size = kernel_size))) ** 0.25
bdata = (intensity(medfilt(data["BR"], kernel_size = kernel_size)) / intensity(medfilt(data["BL"], kernel_size = kernel_size))) ** 0.25

plt.plot(data["time"], tdata, label = "Top", linestyle = styles[0], color = colors[0], linewidth = 2)
plt.plot(data["time"], mdata, label = "Middle", linestyle = styles[2], color = colors[2], linewidth = 2)
plt.plot(data["time"], bdata, label = "Bottom", linestyle = styles[4], color = colors[4], linewidth = 2)

plt.xlabel("Time (s)")
plt.ylabel("\u03B5_A/\u03B5_T")
plt.title("Relative Emissivity of Various Points on Sample During Heating")
plt.legend()
plt.show()
```

```python
t_temp = thermal_model(data["time"], *fitted_params["TL"]["params"])
m_temp = thermal_model(data["time"], *fitted_params["ML"]["params"])
b_temp = thermal_model(data["time"], *fitted_params["BL"]["params"])

plt.figure(figsize=(8, 6))
plt.plot(t_temp, tdata, label="Top", color="red")
plt.plot(m_temp, mdata, label="Middle", color="green")
plt.plot(b_temp, bdata, label="Bottom", color="blue")

plt.xlabel("Fitted temperature (°C)")
plt.ylabel("Emissivity ratio")
plt.title("Relative Emissivity vs fitted temperatures")
plt.legend()
plt.grid(True)
plt.show()
```
